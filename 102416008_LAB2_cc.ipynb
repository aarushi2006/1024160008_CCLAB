Question-2: TUPLE


```python
L=[10, 0, 20, 40, 10, 60, 0, 0]
scores=tuple(L)
highest=max(scores)
low=min(scores)
print("highest score is",highest,"with index",scores.index(highest))
print("lowest score is",low,"which occurs",scores.count(low),"times\n")

reverse=list(scores[::-1])
# Tuples are immutable, so they cannot be modified or reversed in place.
print("Reverse:",reverse,"\n")


num=int(input("enter number to be searched:"))
flag=False
for x in scores:
  if num==x:
    print(num,"occurs at",scores.index(x),"\n")
    flag=True
if flag==False:
  print(num,"NOT FOUND\n")


try:
    scores[0] = 100
except TypeError as e:
    print("Error:", e)
    print("Explanation: Tuples are immutable and cannot be changed after creation, unlike lists.\n")



first_score, second_score, *remaining_scores = scores

print("First score:", first_score)
print("Second score:", second_score)
print("Remaining scores:", remaining_scores)
```

```python
import random

random.seed(1024160008)
numbers=[random.randint(100,900 )for i in range(100)]
print(numbers,"\n")

print("odd numbers are:")
odd=0
for num in numbers:
  if num%2!=0:
    odd=odd+1
    print(num," ",end="")
print("count of odd numbers:",odd,"\n")


print("even numbers are:")
even=0
for num in numbers:
  if num%2==0:
    print(num," ",end="")
    even=even+1
print("count of even numbers:",even,"\n")


def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_list = [num for num in numbers if is_prime(num)]

print("Number of prime numbers:", len(prime_list))
print("Prime numbers:", prime_list,"\n")

most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("Most frequent number:", most_frequent)
print("Frequency:", frequency)
```

```python
# First 8 digits of the roll number
digits = [1, 0, 2, 4, 1, 6, 0, 0]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

print("\nUnion of A and B:", A.union(B))

print("Intersection of A and B:", A.intersection(B))

print("A - B:", A.difference(B))
print("B - A:", B.difference(A))

print("Difference returns elements present in one set but not the other.")
print("Symmetric difference returns elements present in either set but not in both.")

print("Symmetric Difference:", A.symmetric_difference(B))

print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

x = int(input("\nEnter a value to remove from Set A: "))

A.discard(x)
print("Set A after discard:", A)

print("discard() is safer because it does not raise an error if the element is not present, whereas remove() raises a KeyError.")
```

```python
my_dict = {
    "name": "Aarushi Gahlawat",
    "roll_no": "1024160008",
    "branch": "Computer Engineering",
    "age": 20,
    "city": "Gurugram"
}

my_dict["location"] = my_dict.pop("city")
print("After renaming city to location:")
print(my_dict)

my_dict["cgpa"] = 8.88
print("\nAfter adding CGPA:")
print(my_dict)

my_dict["age"] += 1
print("\nAfter increasing age by 1:")
print(my_dict)

dict1 = my_dict.copy()
removed = dict1.pop("branch")
print("\nAfter pop():")
print(dict1)
print("Value returned by pop():", removed)

dict2 = my_dict.copy()
del dict2["branch"]
print("\nAfter del:")
print(dict2)
print("del removes the key without returning its value, while pop() removes and returns the value.")

print("\nDictionary Items:")
for key, value in my_dict.items():
    print(key, "->", value)

key = "email"
if key in my_dict:
    print("\nEmail:", my_dict[key])
else:
    print("\nEmail key not found.")

friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": "1024160015",
    "branch": "Mechanical Engineering",
    "age": 21,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print("\nMerged Dictionary:")
print(merged_dict)
print("When keys are common, the values from the second dictionary overwrite the first.")

string_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("\nDictionary with only string values:")
print(string_dict)
```
