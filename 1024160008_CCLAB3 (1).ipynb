```python
#question4
import kagglehub
import pandas as pd
import os # Import the os module to handle file paths

# Download latest version
path = kagglehub.dataset_download("uciml/iris")

print("Path to dataset files:", path)

# Construct the full path to the CSV file within the downloaded directory
csv_file_path = os.path.join(path, "Iris.csv") # Assuming the CSV file is named 'Iris.csv'

df_iris=pd.read_csv(csv_file_path)
print(df_iris)
print()

#question5
df_iris=df_iris.drop(index=4)
df_iris=df_iris.drop(columns=df_iris.columns[3])
print(df_iris)
print()

#question6
employees = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],"Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
#create dataframe
df=pd.DataFrame(employees)
df.to_csv("employees.csv",index=False)
print("employee dataset created successfully")
print(df)
print()

#find the shape of dataframe
print("shape of dataframe is ", df.shape)
print()

#summary of df
print("summary:",df.info())
print()

#descriptive stats of df
print("Descriptive stats:",df.describe())
print()

#display first 5 and last 3 rows
print("First 5 rows:")
print(df.head(5))
print()
print("Last 3 rows:")
print(df.tail(3))
print()

#calculate required stats
average_salary = df["Salary"].mean()
total_bonus = df["Bonus"].sum()
print(" Total Bonus:",total_bonus)
youngest_age = df["Age"].min()
print("\n(e)(iii) Youngest Employee's Age:",youngest_age)
highest_rating = df["Rating"].max()

print("\n(e)(iv) Highest Performance Rating:")
print(highest_rating)


# --------------------------------------------------
# f) SORT DATAFRAME BY SALARY IN DESCENDING ORDER
# --------------------------------------------------

print("\n(f) DataFrame sorted by Salary in descending order:")

sorted_df = df.sort_values(
    by="Salary",
    ascending=False
)

print(sorted_df)


# --------------------------------------------------
# g) ADD PERFORMANCE CATEGORY
# --------------------------------------------------

def performance_category(rating):

    if rating >= 4.5:
        return "Excellent"

    elif rating >= 4.0:
        return "Good"

    else:
        return "Average"


df["Performance"] = df["Rating"].apply(performance_category)

print("\n(g) Performance Categories:")

print(df[["Name", "Rating", "Performance"]])


# --------------------------------------------------
# h) IDENTIFY MISSING VALUES
# --------------------------------------------------

print("\n(h) Missing Values:")

print(df.isnull().sum())


# --------------------------------------------------
# i) RENAME Employee_ID TO ID
# --------------------------------------------------

df.rename(
    columns={"Employee_ID": "ID"},
    inplace=True
)

print("\n(i) DataFrame after renaming Employee_ID to ID:")

print(df)


# --------------------------------------------------
# j) FIND EMPLOYEES BASED ON CONDITIONS
# --------------------------------------------------

# i. Employees with more than 5 years of experience

print("\n(j)(i) Employees with more than 5 years of experience:")

experienced_employees = df[
    df["Years_of_Experience"] > 5
]

print(experienced_employees)


# ii. Employees belonging to IT department

print("\n(j)(ii) Employees belonging to IT department:")

it_employees = df[
    df["Department"] == "IT"
]

print(it_employees)


# --------------------------------------------------
# k) ADD TAX COLUMN
# --------------------------------------------------

df["Tax"] = df["Salary"] * 0.10

print("\n(k) DataFrame after adding Tax column:")

print(df)


# --------------------------------------------------
# l) SAVE MODIFIED DATAFRAME TO NEW CSV FILE
# --------------------------------------------------

df.to_csv(
    "employees_modified.csv",
    index=False
)

print("\n(l) Modified DataFrame saved successfully.")

print("File name: employees_modified.csv")
```
