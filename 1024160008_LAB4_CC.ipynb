```python
import pandas as pd

# Q1: Create the 4 fixed FAQ entries
fixed_entries = [
    {
        "question": "what is the annual fee",
        "answer": "The annual fee is Rs 500.",
        "keywords": "fee cost price charge",
        "category": "billing"
    },
    {
        "question": "how to reset password",
        "answer": "Go to Settings > Reset Password.",
        "keywords": "password reset login",
        "category": "account"
    },
    {
        "question": "what are your working hours",
        "answer": "We are open 9 AM to 5 PM.",
        "keywords": "hours timing open time",
        "category": "general"
    },
    {
        "question": "how can i pay the fee",
        "answer": "You can pay via UPI, card, or net banking.",
        "keywords": "pay payment upi fee",
        "category": "billing"
    }
]

# Last two digits of roll number 1024160008 are 0 and 8

# Digit 0 -> 0 % 3 = 0 -> billing
entry1 = {
    "question": "how can i check my latest bill",
    "answer": "You can view your latest bill in the billing section of your account.",
    "keywords": "bill billing statement",
    "category": "billing"
}

# Digit 8 -> 8 % 3 = 2 -> general
entry2 = {
    "question": "how can i contact customer support",
    "answer": "You can contact customer support through the help center.",
    "keywords": "support help contact",
    "category": "general"
}

# Add the 2 personalized entries
fixed_entries.append(entry1)
fixed_entries.append(entry2)

# Create the DataFrame
df = pd.DataFrame(fixed_entries)

# Print the final 6-row DataFrame
print("Q1: Final DataFrame")
print(df)


# Q2: Score a query

query = input("\nEnter your FAQ query: ")

# Convert query into individual words
query_words = query.lower().split()

# Store matching results
results = []

# Check every FAQ
for i in range(len(df)):

    # Get keywords of current FAQ
    keywords = df.loc[i, "keywords"].lower().split()

    # Start score at 0
    score = 0

    # Compare query words with FAQ keywords
    for word in query_words:
        if word in keywords:
            score = score + 1

    # Store FAQ if at least one keyword matches
    if score > 0:
        results.append((score, df.loc[i, "question"], df.loc[i, "answer"], df.loc[i, "category"]))

# Sort results according to score
results.sort(reverse=True)

# Print matching results
print("\nQ2: Matching FAQs")

for result in results:
    print("Score:", result[0])
    print("Question:", result[1])
    print("Answer:", result[2])
    print("Category:", result[3])
    print()


# Q3: Find FAQs belonging to a category

category_name = "billing"

print("Q3: FAQs in", category_name, "category")

print(df[df["category"] == category_name])


# Q4: Add a new keyword to one FAQ

print("\nQ4: Add a new keyword")

# Display the FAQ numbers
print(df[["question"]])

# Ask which FAQ to update
entry_number = int(input("Enter the row number to update: "))

# Ask for the new keyword
new_keyword = input("Enter a new keyword: ")

# Add the new keyword
df.loc[entry_number, "keywords"] = df.loc[entry_number, "keywords"] + " " + new_keyword

# Save the updated DataFrame
df.to_csv("1024160008_faq_data.csv", index=False)

print("\nUpdated DataFrame:")
print(df)

print("\nFile saved as 1024160008_faq_data.csv")


# Q5: Count FAQ entries in each category

print("\nQ5: Number of FAQs in each category")

print(df.groupby("category").size())


# Q6: Handle ties in scoring

# Example query that produces a tie
query = "fee"

# Split query into words
query_words = query.lower().split()

# Store results
results = []

# Check every FAQ
for i in range(len(df)):

    keywords = df.loc[i, "keywords"].lower().split()

    score = 0

    # Calculate score
    for word in query_words:
        if word in keywords:
            score = score + 1

    # Store matching FAQ
    if score > 0:
        results.append((score, df.loc[i, "question"], df.loc[i, "answer"], df.loc[i, "category"]))

# Find highest score
highest_score = max(result[0] for result in results)

print("\nQ6: Query =", query)
print("Highest score =", highest_score)

# Print all FAQs having the highest score
for result in results:
    if result[0] == highest_score:
        print("Score:", result[0])
        print("Question:", result[1])
        print("Answer:", result[2])
        print("Category:", result[3])
        print()
```
